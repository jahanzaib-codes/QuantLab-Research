# Statistical Arbitrage: Cointegration & Pairs Trading Engine

This research notebook demonstrates an institutional implementation of the **Engle-Granger Two-Step Cointegration Test**, **Ordinary Least Squares (OLS) Hedge Ratio estimation**, and **Ornstein-Uhlenbeck Half-Life** mean-reversion modeling on equities/crypto pairs.

```python
import numpy as np
import pandas as pd
from src.stats.stat_arb import test_cointegration
from src.backtest.engine import VectorizedBacktester
from cli import generate_benchmark_pair_data

# Generate 250 bars of cointegrated pair series (NVDA and AMD)
nvda, amd = generate_benchmark_pair_data(250)
coint = test_cointegration(nvda, amd, 'NVDA', 'AMD')

print(f"Hedge Ratio (Beta): {coint.hedge_ratio}")
print(f"ADF Statistic: {coint.adf_statistic} (95% Critical Value: {coint.critical_values['95%']})")
print(f"Is Cointegrated: {coint.is_cointegrated_95pct}")
print(f"Ornstein-Uhlenbeck Half-Life: {coint.half_life_days} days")
print(f"Current Spread Z-Score: {coint.current_zscore} sigma")
```

```python
# Execute vectorized backtest
backtester = VectorizedBacktester(initial_capital=100000.0, slippage_bps=3.0)
tearsheet = backtester.evaluate_pairs_strategy(nvda, amd, coint.hedge_ratio)

print(f"CAGR: {tearsheet.cagr_pct}%")
print(f"Sharpe Ratio: {tearsheet.sharpe_ratio}")
print(f"Max Drawdown: {tearsheet.max_drawdown_pct}%")
print(f"Win Rate: {tearsheet.win_rate_pct}%")
```
